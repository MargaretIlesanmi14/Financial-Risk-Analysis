```python
# Load the CSV file into a pandas DataFrame
df = pd.read_csv('financial_risk_dataset.csv')

# Display the first 5 rows to ensure it loaded correctly
df.head()
```

```python
# Check column names, data types, and missing values
df.info()
```

```python
# Check summary statistics of numeric and categorical columns
df.describe(include='all')
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Fill missing values for numerical columns
df['Coapplicant_Income'] = df['Coapplicant_Income'].fillna(df['Coapplicant_Income'].median())
df['Collateral_Value'] = df['Collateral_Value'].fillna(df['Collateral_Value'].median())
df['Credit_Utilization_Ratio'] = df['Credit_Utilization_Ratio'].fillna(df['Credit_Utilization_Ratio'].median())
```

```python
# Check missing values again
df.isnull().sum()
```

# Missing values were found in Coapplicant_Income, Collateral_Value, and Credit_Utilization_Ratio.
# We filled numerical missing values with the median to avoid skewing the data.
# All other columns were complete and required no changes.


```python
# Convert date columns to datetime type
df['Loan_Start_Date'] = pd.to_datetime(df['Loan_Start_Date'])
df['Loan_End_Date'] = pd.to_datetime(df['Loan_End_Date'])
```

```python
# Ensure ratio/percentage columns are float (BI-ready)
ratio_cols = [
    'Credit_Utilization_Ratio',
    'Debt_to_Income_Ratio',
    'Interest_Rate',
    'Probability_of_Default'
]

for col in ratio_cols:
    df[col] = df[col].astype(float)
```

```python
# Identify duplicate rows
df.duplicated().sum()

# Remove duplicates if any exist
df = df.drop_duplicates()
```

```python
# To check for outliers, select numeric columns (int64 and float64)
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
print(num_cols)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# List of numeric columns to check for outliers
num_cols = ['Applicant_Age', 'Years_in_Employment', 'Dependents',
            'Applicant_Income', 'Coapplicant_Income', 'Credit_Score',
            'Existing_Debt', 'Loan_Amount', 'Loan_Term_Months', 'Interest_Rate',
            'Collateral_Value', 'Payment_Delays_6mo', 'Credit_Utilization_Ratio',
            'Debt_to_Income_Ratio', 'Probability_of_Default', 'Default_Status']

plt.figure(figsize=(18, 12))
for i, col in enumerate(num_cols):
    plt.subplot(4, 4, i+1)
    sns.boxplot(y=df[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
# Check negative Interest Rate values
negative_interest = df[df['Interest_Rate'] < 0]
print("Negative Interest Rate rows:")
print(negative_interest)
```

```python
# Fix: Set negative interest rates to NaN or take absolute value if it's a data entry error
df['Interest_Rate'] = df['Interest_Rate'].apply(lambda x: abs(x) if x < 0 else x)
```

```python
# Check Credit Utilization Ratio > 100%
high_credit_util = df[df['Credit_Utilization_Ratio'] > 100]
print("\nCredit Utilization Ratio > 100% rows:")
print(high_credit_util)
```

```python
# Fix: Cap values at 100%
df['Credit_Utilization_Ratio'] = df['Credit_Utilization_Ratio'].apply(lambda x: 100 if x > 100 else x)
```

```python
# Quick summary after fixes
print("\nSummary after error correction:")
print(df[['Interest_Rate', 'Credit_Utilization_Ratio']].describe())
```

```python
# Export cleaned dataset for Power BI
df.to_csv('financial_risk_cleaned.csv', index=False)
```
